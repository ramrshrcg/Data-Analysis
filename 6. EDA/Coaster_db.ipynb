```python
#since some columns hace too many missing values, we will drop those columns
columns_to_drop= ['Gforce_clean','height_ft','Fast Lane available','Fastrack available','Opened','Replaced by','Website','Flash Pass Available','Must transfer from wheelchair','Theme','Single rider line available','Restraint Style','Flash Pass available', 'Acceleration', 'Restraints', 'Name','year_introduced']
df_clean.drop(columns= columns_to_drop, inplace= True)
print(df_clean.shape)
df_clean.head(5)
```

```python
df_clean.isna().sum()
```

```python
columns_to_drop2= ['Cost','Park section','G-force','Max vertical angle','Soft opening date','Replaced','Track layout', 'Soft opening date.1', 'Closing date']
df_clean.drop(columns=columns_to_drop2, inplace = True)

```

```python
df_copy = df[['coaster_name',
    # 'Length', 'Speed',
    'Location', 'Status',
    # 'Opening date',
    #   'Type',
    'Manufacturer',
#     'Height restriction', 'Model', 'Height',
#        'Inversions', 'Lift/launch system', 'Cost', 'Trains', 'Park section',
#        'Duration', 'Capacity', 'G-force', 'Designer', 'Max vertical angle',
#        'Drop', 'Soft opening date', 'Fast Lane available', 'Replaced',
#        'Track layout', 'Fastrack available', 'Soft opening date.1',
#        'Closing date',
#     'Opened', 
    # 'Replaced by', 'Website',
#        'Flash Pass Available', 'Must transfer from wheelchair', 'Theme',
#        'Single rider line available', 'Restraint Style',
#        'Flash Pass available', 'Acceleration', 'Restraints', 'Name',
       'year_introduced',
        'latitude', 'longitude',
    'Type_Main',
       'opening_date_clean',
    #'speed1', 'speed2', 'speed1_value', 'speed1_unit',
       'speed_mph', 
    #'height_value', 'height_unit',
    'height_ft',
       'Inversions_clean', 'Gforce_clean']].copy()
```

```python
df_copy.head(15)
```

```python
df_copy.info()
```

```python
df_copy['opening_date_clean']= pd.to_datetime(df_copy['opening_date_clean'])
```

```python
# Rename our columns
df_copy = df_copy.rename(columns={'coaster_name':'Coaster_Name',
                   'year_introduced':'Year_Introduced',
                   'opening_date_clean':'Opening_Date',
                   'speed_mph':'Speed_mph',
                   'height_ft':'Height_ft',
                   'Inversions_clean':'Inversions',
                   'Gforce_clean':'Gforce'})
```

```python
df_copy = df_copy.loc[~df_copy.duplicated(subset=['Coaster_Name','Location','Opening_Date'])].reset_index(drop=True).copy()
```

```python
top10data= df_copy['Year_Introduced'].value_counts().head(10)
ax= top10data.plot(kind='bar', title='Top 10 Years with Most Coaster Introductions')
ax.set_xlabel('Year Introduced')
ax.set_ylabel('Number of Coasters Introduced')
```

```python
plt.figure(figsize=(12,6))
plt.bar(top10data.index, top10data.values)
plt.xlabel('Year Introduced')
plt.ylabel('Number of Coasters Introduced')
plt.title('Number of Coasters Introduced Each Year')
plt.show()
```
